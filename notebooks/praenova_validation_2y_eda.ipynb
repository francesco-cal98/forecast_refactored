# Praenova Validation 2Y - Per-Series EDA

This notebook treats each `(product, area)` as an independent time series and builds combined dashboards for cross-series comparison.

```python
from pathlib import Path
import sys
import pandas as pd

# Make notebook runnable both from repo root and from notebooks/ as cwd.
if (Path.cwd() / 'backend').exists():
    PROJECT_ROOT = Path.cwd()
elif (Path.cwd().parent / 'backend').exists():
    PROJECT_ROOT = Path.cwd().parent
else:
    raise RuntimeError('Could not find project root containing backend/.')

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from backend.prototype.praenova.io import load_area_panel
from backend.prototype.praenova.quality import run_quality_checks
from backend.prototype.praenova.summary import build_series_summary
from backend.prototype.praenova.run_eda import run_eda
```

```python
INPUT_FP = PROJECT_ROOT / 'praenova_validation_2y.xlsx'
OUTPUT_DIR = PROJECT_ROOT / 'outputs' / 'eda' / 'praenova_validation_2y'

paths = run_eda(input_path=str(INPUT_FP), output_dir=str(OUTPUT_DIR), base_year=2023)
paths
```

```python
df = load_area_panel(str(INPUT_FP), base_year=2023)
quality = run_quality_checks(df)
summary = build_series_summary(df)

print('Rows:', len(df))
print('Series:', df['series_id'].nunique())
print('Duplicate keys:', quality['duplicate_key_count'])
summary.head()
```

```python
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

dashboard_files = [
    'dashboard_trend.png',
    'dashboard_seasonality.png',
    'dashboard_promo_impact.png',
    'dashboard_price_vs_sales.png',
]

for name in dashboard_files:
    fp = OUTPUT_DIR / name
    img = mpimg.imread(fp)
    plt.figure(figsize=(16, 9))
    plt.imshow(img)
    plt.title(name)
    plt.axis('off')
    plt.show()
```
